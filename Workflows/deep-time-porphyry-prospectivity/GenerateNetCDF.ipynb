```python
from pyDTDM import *
import pandas as pd
```

```python
NETCDF_GRID_RESOLUTION=0.5
```

```python
prediction_data=pd.read_csv(f"<DATA_ROOT>/CopperLithium/NorthAmerica/output/outputs.csv")
```

```python
create_directory_if_not_exists(f"<DATA_ROOT>/CopperLithium/NorthAmerica/output/probability_grids")

for reconstruction_time in prediction_data['age (Ma)'].unique():
    prediction_data_t=prediction_data[prediction_data['age (Ma)']==reconstruction_time].copy()
    da = df_to_NetCDF(
        x=prediction_data_t['Longitude'],
        y=prediction_data_t['Latitude'],
        z=prediction_data_t[f"Probability(1)"],
        statistic='mean',
        grid_resolution=NETCDF_GRID_RESOLUTION,
        clip=(None, None),
        lat_bin_edges=np.arange(-90, 90 + NETCDF_GRID_RESOLUTION, NETCDF_GRID_RESOLUTION),
        lon_bin_edges=np.arange(-180,180 + NETCDF_GRID_RESOLUTION, NETCDF_GRID_RESOLUTION))
    da.to_netcdf(f"<DATA_ROOT>/CopperLithium/NorthAmerica/output/probability_grids/grids_{reconstruction_time}.nc")
```

```python
reconstruction_time=25
prediction_data_t=prediction_data[prediction_data['age (Ma)']==reconstruction_time].copy()
da = df_to_NetCDF(
    x=prediction_data_t['Longitude'],
    y=prediction_data_t['Latitude'],
    z=prediction_data_t[f"Probability(1)"],
    statistic='mean',
    grid_resolution=NETCDF_GRID_RESOLUTION,
    clip=(None, None),
    lat_bin_edges=np.arange(-90, 90 + NETCDF_GRID_RESOLUTION, NETCDF_GRID_RESOLUTION),
    lon_bin_edges=np.arange(-180,180 + NETCDF_GRID_RESOLUTION, NETCDF_GRID_RESOLUTION))
da.plot()
```
